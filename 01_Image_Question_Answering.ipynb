# 1. Ask a question about a Texas image

**Idea:** give a vision-language model an image and a question.
This example uses one NOAA image of Bear Creek, west Houston, during the
Hurricane Harvey response. We provide no labeled examples: this is zero-shot.

Open this notebook from this folder. See `README.txt` for installation.
The reusable functions are in [texas_flood.py](texas_flood.py);
your model connection is in [vlm_config.py](vlm_config.py).


```python
from texas_flood import load_texas_images, show_images, ask_image

image = load_texas_images(["TX001_2017_NOAA"])
show_images(image)
```

## Ask one question

`ask_image` sends the image and question to Qwen and returns its answer.
An ordinary river or pond alone is not evidence of flooding.


```python
question = "Is floodwater visibly covering normally dry land? Answer yes, no, or uncertain and give one visible clue."
answer = ask_image(image.iloc[0]["path"], question)
print(answer)
```
